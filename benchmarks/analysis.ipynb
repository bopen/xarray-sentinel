# Benchmark results

Median wall time of the successful runs in `results/*.jsonl`.

```python
import json
from pathlib import Path

import pandas as pd

records = []
for path in Path("results").glob("*.jsonl"):
    records += [json.loads(line) for line in path.read_text().splitlines()]
df = pd.json_normalize(records)
df = df[df.returncode == 0]
df["product"] = df.source.str.rstrip("/").str.rsplit("/", n=1).str[-1]
df["operation"] = df.get("operation", "convert").fillna("convert")
```

```python
keys = ["label", "operation", "product", "tool"]
summary = df.groupby(keys).wall_time.median().unstack("tool")
summary
```

```python
summary.plot.barh(xlabel="wall time [s]", figsize=(8, 1 + len(summary) / 2))
```
